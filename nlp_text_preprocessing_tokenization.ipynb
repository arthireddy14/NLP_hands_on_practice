# Text Preprocessing in NLP

Before training NLP models, we must clean the text.

This process is called Text Preprocessing.

# Tokenization

Breaking text into smaller units (tokens).

It involves dividing a Textual input into smaller units known as tokens. These tokens can be in the form of words, characters, sub-words, or sentences. It helps in improving interpretability of text by different models.

```python
paragraph = """
Natural Language Processing (NLP) is a field of Artificial Intelligence.
It helps computers understand human language.
Tokenization is an important NLP technique.
"""
```

```python
from nltk.tokenize import sent_tokenize
sentences=sent_tokenize(paragraph)
print(sentences)
```

```python
from nltk.tokenize import word_tokenize
words=word_tokenize(paragraph)
print(words)
```

```python
tot_sentences=len(sentences)
tot_words=len(words)
unique_words=len(set(words))
print("Total Sentences:",tot_sentences)
print("Total Words:",tot_words)
print("Unique Words:",unique_words)
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
stop_words=set(stopwords.words('english'))
tokens=word_tokenize(paragraph)
filtered_words=[w for w in tokens if w.lower() not in stop_words]
print("Original paragraph :")
print(paragraph)
print("After Stopword Removal : ")
print(filtered_words)
```

```python
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
stemmer=PorterStemmer()
lemmatizer=WordNetLemmatizer()
stemmed=[stemmer.stem(word) for word in filtered_words]
lemmatized=[lemmatizer.lemmatize(word) for word in filtered_words]
print("Stemmed words :")
print(stemmed)
print("Lemmatized words :")
print(lemmatized)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
vectorizer=CountVectorizer()
documents=["I love NLP","NLP is Amazing"]
x=vectorizer.fit_transform(documents)
print("Vocabulary :")
print(vectorizer.get_feature_names_out())
print("Documnet vectors :")
print(x.toarray())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

documents = [
    "NLP is amazing",
    "NLP is powerful",
    "Machine learning is powerful"
]

tfidf = TfidfVectorizer()

matrix = tfidf.fit_transform(documents)

feature_names = tfidf.get_feature_names_out()

scores = matrix.sum(axis=0).A1

tfidf_scores = pd.DataFrame({
    'Word': feature_names,
    'Score': scores
})

tfidf_scores = tfidf_scores.sort_values(
    by='Score',
    ascending=False
)

print("Top Important Words:")
print(tfidf_scores)
```
